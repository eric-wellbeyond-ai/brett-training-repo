## 1. Run simple, raw Python

A notebook is a Python interpreter with notes in between. The next cells are ordinary Python — no files, no libraries, no classes yet.

````python
print("Hello, Brett.")
print(2 + 2)
print(3 * 7)
print(2 ** 8)  # ** means exponent: 2 to the 8th power
````

**Variables** store values under a name. Python figures out the type for you.

````python
name = "Brett"
games = 10
points = 187
ppg = points / games

print(name)
print(type(name), type(games), type(ppg))
print(f"{name} averaged {ppg:.1f} points per game.")
````

**Lists** hold an ordered collection. **`for`** loops walk through them. **`if`** chooses a branch.

````python
scores = [22, 31, 18, 27, 14]

total = 0
for score in scores:
    total = total + score

average = total / len(scores)
print("scores:", scores)
print("average:", average)

if average >= 20:
    print("Hot streak.")
else:
    print("Cold night.")
````

A **function** packages a piece of logic you can reuse. `def` defines it; calling `double(5)` runs it.

````python
def double(x):
    return x * 2

def greeting(person):
    return f"Welcome to Python, {person}."

print(double(5))
print(greeting("Brett"))
````

## 2. Load a class from a `.py` file

Object-oriented programming (OOP) groups *data* and *behavior* together.

- A **class** is the blueprint (`Player` in `player.py`).
- An **instance** (or object) is one concrete thing built from that blueprint (`lebron`, `jokic`).
- A **property** (attribute) is data stored on the instance: `lebron.team`.
- A **method** is a function that belongs to the class: `lebron.points_per_game()`.

Open `player.py` in the editor and read it, then run the cells below.

### 2a. Instantiate the class

`from player import Player` loads the class defined in `player.py` in this folder. Calling `Player(...)` **instantiates** it — it runs `__init__` and returns a new object.

````python
from player import Player

lebron = Player("LeBron James", "Lakers", 1766, 71)
jokic = Player("Nikola Jokic", "Nuggets", 2094, 79)

print(lebron)
print(type(lebron))
print(jokic)
````

### 2b. Access a property

Dot notation reads data stored on the instance. `lebron.name` and `lebron.team` are properties set in `__init__`.

````python
print(lebron.name)
print(lebron.team)
print(lebron.points)
print(lebron.games)

# You can also assign a new value to a property.
lebron.team = "Lakers"
print("team is still:", lebron.team)
````

### 2c. Access a method

A method is called with parentheses. `self` inside the class is the instance you called it on, so `lebron.points_per_game()` uses LeBron's points and games.

````python
print(lebron.points_per_game())
print(jokic.points_per_game())
print(lebron.summary())
print(jokic.summary())
````

### `numpy` — arrays and numeric math

A NumPy **array** is like a list, but every element is the same type and math is vectorized: you operate on the whole array at once instead of writing a `for` loop.

````python
import numpy as np

game_points = np.array([22, 31, 18, 27, 14, 29, 33])

print("array:", game_points)
print("mean:", game_points.mean())
print("max:", game_points.max())
print("min:", game_points.min())
print("plus 2 (every game):", game_points + 2)
print("shooting a 3-game rolling-style slice [1:4]:", game_points[1:4])

# A 2-D array: 3 players x 4 games
box = np.array(
    [
        [22, 31, 18, 27],
        [14, 29, 33, 21],
        [19, 16, 25, 30],
    ]
)
print("shape (rows, cols):", box.shape)
print("per-player totals:", box.sum(axis=1))
print("per-game totals:", box.sum(axis=0))
````

## 4. Interoperability with Wolfram Language

Python and Wolfram Language can call each other. That is useful when you want Python's data libraries *and* Wolfram's math, entities, and notebooks.

There are two directions:

| Direction | Where you write it | Tool |
| --- | --- | --- |
| **Wolfram runs inside Python** | this notebook | `wolframclient` |
| **Python runs inside Wolfram** | `python_from_wolfram.vsnb` | `ExternalEvaluate` |

Open **`python_from_wolfram.vsnb`** for the second direction (section 4b).

### 4a. Run Wolfram Language from Python

The [Wolfram Client Library](https://reference.wolfram.com/language/WolframClientForPython/) starts a local Wolfram kernel and evaluates expressions.

You need Wolfram Engine or Mathematica installed. The next cell tries the usual macOS kernel paths, then falls back to auto-discovery. If it still cannot start a kernel, it prints a short note instead of crashing the rest of the notebook.

````python
from wolframclient.evaluation import WolframLanguageSession
from wolframclient.language import wl, wlexpr

# wolframclient looks for a kernel on the PATH. On macOS it is often
# easier to pass the path explicitly. Add your own path if needed.
kernel_candidates = [
    "/Applications/Wolfram.app/Contents/MacOS/WolframKernel",
    "/Applications/Wolfram Engine.app/Contents/MacOS/WolframKernel",
    "/Applications/Mathematica.app/Contents/MacOS/WolframKernel",
    None,  # last resort: auto-discover
]

session = None
last_error = None
for kernel in kernel_candidates:
    try:
        session = WolframLanguageSession(kernel) if kernel else WolframLanguageSession()
        session.start()
        break
    except Exception as exc:
        last_error = exc
        session = None

if session is None:
    print("Could not start a Wolfram kernel from Python.")
    print("Install Wolfram Engine (or Mathematica) and try again.")
    print("Details:", type(last_error).__name__, last_error)
else:
    try:
        # Evaluate a Wolfram Language string.
        squares = session.evaluate(wlexpr("Map[#^2 &, Range[5]]"))
        print("squares of 1..5:", squares)

        # Call a Wolfram function as if it were Python: wl.FunctionName(...)
        pi_approx = session.evaluate(wl.N(wl.Pi, 20))
        print("Pi to 20 digits:", pi_approx)

        stats = session.evaluate(wl.Mean([22, 31, 18, 27, 14]))
        print("Wolfram Mean of scores:", stats)
    finally:
        session.terminate()
        print("Wolfram session closed.")
````

## 5. Python for sports statistics

Python is the default toolkit for sports analytics because the whole pipeline lives in one language. You pull a season's box scores from a public API or a downloadable CSV, land them in a pandas DataFrame, then slice, group, and model. The same session can plot a shooting chart, fit a simple expected-points model, or feed a machine-learning library. Analysts at teams, sportsbooks, and sites like FiveThirtyEight and Basketball-Reference-style blogs all work this way: get the data, tidy it, then ask questions the box score does not answer on its own — true shooting percentage, plus-minus, pace, strength of schedule.

You do not need a paid feed to start. Packages such as `nba_api` wrap NBA.com endpoints; college and soccer communities maintain similar clients; many leagues publish CSVs. Once the table is in pandas, the analysis is ordinary data work: `groupby` for team totals, new columns for rates (points per game, assist-to-turnover), filters for home vs away, and `numpy` for the numeric guts. The cells below stay offline on purpose — they use `sample_nba_stats.csv` in this folder — but the pattern is the same if you later replace the CSV with a live API call.

**A typical next step** would look like `from nba_api.stats.endpoints import playergamelog` (or an equivalent for your sport), convert the result with `.get_data_frames()[0]`, then compute the same PPG / RPG / APG columns you see below.

````python
import pandas as pd
from player import Player

stats = pd.read_csv("sample_nba_stats.csv")

# Rate stats: totals become per-game numbers.
stats["ppg"] = stats["points"] / stats["games"]
stats["rpg"] = stats["rebounds"] / stats["games"]
stats["apg"] = stats["assists"] / stats["games"]

print("Scoring leaders (PPG)")
print(stats.sort_values("ppg", ascending=False)[["player", "team", "ppg", "rpg", "apg"]].round(1))
print()
print("Team scoring (total points)")
print(stats.groupby("team")["points"].sum().sort_values(ascending=False))
````

````python
# Same data, now as Player objects from player.py.
# This is how OOP and pandas meet: the table is for analysis,
# the class is for modeling one athlete at a time.
roster = [
    Player(row.player, row.team, row.points, row.games)
    for row in stats.itertuples()
]

for athlete in roster:
    print(athlete.summary())

best = max(roster, key=lambda p: p.points_per_game())
print()
print("Best PPG on this sheet:", best.name, round(best.points_per_game(), 1))
````
